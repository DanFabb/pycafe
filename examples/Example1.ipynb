```python
import pycafe
```

```python
# Import fluid properties
fluid = pycafe.load_fluid("air")

```

```python

# Import mesh
nodes, elements, boundaries = pycafe.load_mesh("rectangle_CQUAD8.msh")

```

```python

# Assign boundary conditions
bc = pycafe.assign_boundary_conditions(
    boundaries=boundaries,
    nodes=nodes,
    interactive=False,
)
```

```python

# % Prepare the acoustic system
system = pycafe.prepare_acoustic_system(
    nodes=nodes,
    elements=elements,
    boundaries=boundaries,
    rho=fluid["rho"],
    c0=fluid["c0"],
    bc=bc,
    debug=True,
)

K_red = system["K_red"]
M_red = system["M_red"]
C_red = system["C_red"]





```

```python

#  Run the analysis and to the post processing needed
results = pycafe.run_analysis(
    system=system,
    nodes=nodes,
    boundaries=boundaries,
    rho=fluid["rho"],
    interactive=True,
)

```
