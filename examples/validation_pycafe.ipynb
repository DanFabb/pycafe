# pyCAFE - Validation Benchmark

This notebook is the publication-facing companion to `validation_pycafe.py`.

It runs the full 2D rectangular-cavity validation workflow and generates:
- modal accuracy comparisons against analytical solutions,
- an `h`-refinement study for `CQUAD4` and `CQUAD8`,
- pyCAFE versus FEniCSx modal comparisons,
- direct-frequency-response comparisons,
- figure files saved next to this notebook.

The benchmark covers both hard-wall (Neumann) and pressure-release (`p=0`) cavity configurations.

## Requirements

Run this notebook in an environment that already contains:
- `pycafe`
- `gmsh`
- `numpy`
- `matplotlib`
- `mpi4py`
- `dolfinx`
- `ufl`
- `scipy`
- `slepc4py`

The script is executed as-is, so any figures generated by a previous run in the same folder will be overwritten.

```python
import pathlib
import runpy

workdir = pathlib.Path.cwd()
script_path = workdir / "validation_pycafe.py"
print(f"Running benchmark script: {script_path}")

if not script_path.exists():
    raise FileNotFoundError("validation_pycafe.py must be in the current working directory")

runpy.run_path(str(script_path), run_name="__main__")
```
